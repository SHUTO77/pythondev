```python
from string import ascii_lowercase  # 'abcdefghijklmnopqrstuvwxyz'

# 1) 空の英単語リスト
words = []

# 2) a〜z:0 の辞書
counts = {ch: 0 for ch in ascii_lowercase}

# 3) 入力ループ（空行で終了）
while True:
    s = input("英単語を入力してください：").strip()
    if s == "":        # 空行で終了
        break

    words.append(s)    # リストへ保存

    # 文字数カウント（小文字化し、a〜z のみカウント）
    for ch in s.lower():
        if ch in counts:
            counts[ch] += 1

# 4-1) 単語リストを ABC 順で表示
sorted_words = sorted(words)
print(f"入力した英単語：{sorted_words}")

# 4-2) 出現があった文字だけ表示（a→z の順）
for ch in ascii_lowercase:
    if counts[ch] > 0:
        print(f"{ch}が{counts[ch]}個ありました")
```
