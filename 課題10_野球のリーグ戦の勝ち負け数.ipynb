```python
# Baseball_Team クラスを定義
class Baseball_Team:
    def __init__(self, name, win, lose, draw):
        self.name = name   # チーム名
        self.win = win     # 勝ち数
        self.lose = lose   # 負け数
        self.draw = draw   # 引き分け数

    # 勝率を計算するメソッド
    def calc_win_rate(self):
        return self.win / (self.win + self.lose)

    # チームの勝敗情報を表示するメソッド
    def show_team_result(self):
        # f文字列で整形（桁揃え）
        print(f"{self.name:<9}{self.win:>3}{self.lose:>5}{self.draw:>5}{self.calc_win_rate():6.3f}")


# 6チームのインスタンスを生成
teams = [
    Baseball_Team("Archers", 77, 64, 2),
    Baseball_Team("Meteors", 71, 69, 3),
    Baseball_Team("Cheetahs", 69, 68, 6),
    Baseball_Team("Sharks", 70, 70, 3),
    Baseball_Team("Pythons", 68, 73, 2),
    Baseball_Team("Owls", 59, 82, 2),
]

# 見出しを表示
print(f"{'team':<9}{'win':>3}{'lose':>5}{'draw':>5}{'rate':>6}")

# 各チームの結果を表示
for team in teams:
    team.show_team_result()
```
