```python
import numpy as np

x = np.array([62, 95, 83, 100, 72, 71, 55, 85, 67, 75])   # english
y = np.array([62, 80, 85, 95, 58, 76, 52, 84, 67, 71])    # math

import matplotlib.pyplot as plt
import numpy as np

# 最小二乗法で y ≒ m*x + b を当てはめる（m=傾き, b=切片）
m, b = np.polyfit(x, y, 1)

# 直線をきれいに描くため、x を昇順に並べた点で直線のyを作成
x_line = np.sort(x)
y_line = m * x_line + b

# 散布図（青）と近似直線（赤）
plt.scatter(x, y, color='blue', label='score')   # 散布図：青
plt.plot(x_line, y_line, color='red', label='lsm')  # 近似直線：赤

# 軸ラベルと凡例（左上）
plt.xlabel('english')
plt.ylabel('math')
plt.legend(loc='upper left')

plt.show()
```
